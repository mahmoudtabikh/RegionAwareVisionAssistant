```python
# 1. Import required modules
from anomalib.data import MVTecAD
from anomalib.deploy import ExportType
from anomalib.engine import Engine
from anomalib.models import EfficientAd
import torch
```

```python
device = "cuda"  # Use GPU if available, otherwise use CPU
torch.set_float32_matmul_precision('medium')
```

```python
# 2. Create a dataset
# MVTecAD is a popular dataset for anomaly detection
datamodule = MVTecAD(
    root="/home/mahmoud/projects/RegionAwareVisionAssistant/data/MVTecAD",  # Path to download/store the dataset
    category="leather",  # MVTec category to use
    train_batch_size=1,  # Number of images per training batch
    eval_batch_size=1,  # Number of images per validation/test batch
)
```

```python
# 3. Initialize the model
model = EfficientAd(
    imagenet_dir = "/home/mahmoud/projects/RegionAwareVisionAssistant/src/data/MVTecAD",
    model_size="small",
)
```

```python
# 4. Create the training engine
engine = Engine()
```

```python
# 5. Train the model
# This produces a lightning model (.ckpt)
engine.fit(datamodule=datamodule, model=model)
```

```python
# 6. Test the model performance
test_results = engine.test(datamodule=datamodule, model=model)
```

```python
# 7. Export the model
# Different formats are available: Torch, OpenVINO, ONNX
engine.export(
    model=model,
    export_type=ExportType.OPENVINO,
)
```
