```python
import json
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
results_path = "/home/mahmoud/projects/RegionAwareVisionAssistant/results/EfficientAd/MVTecAD/wood/wood_test_scores.json"

with open(results_path, "r") as f:
    results = json.load(f)

test_df = pd.DataFrame(results)
test_df['image_name'] = test_df['image_path'].apply(lambda x: '_'.join(x.split('/')[-3:]))
test_df['anomaly_category'] = test_df['image_name'].apply(lambda x: x.split('_')[1])
test_df.head()
```

```python
best_threshold = 0.5001757740974426
test_df['pred_label_calibrated'] = test_df['pred_score'] >= best_threshold

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, f1_score, precision_score, recall_score

cm = confusion_matrix(test_df['gt_label'], test_df['pred_label_calibrated'])
ConfusionMatrixDisplay(cm, display_labels=['Good', 'Defect']).plot()

print("F1:", f1_score(test_df['gt_label'], test_df['pred_label_calibrated']))
print("Precision:", precision_score(test_df['gt_label'], test_df['pred_label_calibrated']))
print("Recall:", recall_score(test_df['gt_label'], test_df['pred_label_calibrated']))
```

```python
final_metrics = {
    "threshold": float(best_threshold),
    "f1": f1_score(test_df['gt_label'], test_df['pred_label_calibrated']),
    "precision": precision_score(test_df['gt_label'], test_df['pred_label_calibrated']),
    "recall": recall_score(test_df['gt_label'], test_df['pred_label_calibrated']),
    "val_size": 15,
    "test_size": 64,
}
with open("/home/mahmoud/projects/RegionAwareVisionAssistant/results/EfficientAd/MVTecAD/wood/wood_final_metrics.json", "w") as f:
    json.dump(final_metrics, f, indent=2)
```
