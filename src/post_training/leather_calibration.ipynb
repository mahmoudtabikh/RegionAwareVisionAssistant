```python
import json
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
results_path = "/home/mahmoud/projects/RegionAwareVisionAssistant/results/EfficientAd/MVTecAD/leather/leather_val_scores.json"

with open(results_path, "r") as f:
    results = json.load(f)

df = pd.DataFrame(results)
df['image_name'] = df['image_path'].apply(lambda x: '_'.join(x.split('/')[-3:]))
df['anomaly_category'] = df['image_name'].apply(lambda x: x.split('_')[1])
df.head()
```

```python
sns.displot(df, x='pred_score', hue='gt_label', multiple="dodge")
```

```python
sns.displot(df, x='pred_score', hue='anomaly_category', multiple='dodge')
```

```python
sns.stripplot(df, x='pred_score', y='gt_label', jitter=True)
plt.xticks(rotation=45)
```

```python
sns.swarmplot(df, x='pred_score', y='gt_label')
plt.xticks(rotation=45)
```
